## 1. Opening and Closing Files

The built-in `open()` function is your gateway to file operations. It returns a **file object** that you use to read or write data.

```python
open(filename, mode)
```

| Parameter  | Description |
|------------|-------------|
| `filename` | Path to the file (string) |
| `mode`     | How you want to use the file (read, write, etc.) |

**Important:** Every file you open should eventually be closed to free up system resources.

### Creating a File

````python
# Step 1: Open a file in write mode ('w')
# If the file doesn't exist, Python creates it automatically.
# If it already exists, 'w' mode will OVERWRITE the contents.
my_file = open("hello.txt", "w")
````

````python
# --- Creating a file using open() and close() ---

# Step 1: Open a file in write mode ('w')
# If the file doesn't exist, Python creates it automatically.
# If it already exists, 'w' mode will OVERWRITE the contents.
my_file = open("hello.txt", "w")

# Step 2: Write some text into the file
my_file.write("Hello, World!\n")
my_file.write("Welcome to Python file handling.\n")
my_file.write("This is line number three.\n")

# Step 3: Always close the file when you're done
my_file.close()

print("✅ File 'hello.txt' created successfully!")
````

### Reading a File

Now let's read back what we just wrote. There are several ways to read file contents:

````python
# --- Method 1: read() — Reads the ENTIRE file as one big string ---

my_file = open("hello.txt", "r")   # 'r' = read mode (this is the default)
content = my_file.read()            # Read everything at once
my_file.close()                     # Don't forget to close!

print("--- Full file content ---")
print(content)
````

````python
# --- Method 2: readline() — Reads ONE line at a time ---
# Useful when you want to process a file line by line

my_file = open("hello.txt", "r")

first_line = my_file.readline()     # Reads the first line
second_line = my_file.readline()    # Reads the next line

my_file.close()

print("First line: ", first_line, end="")   # end="" avoids double newline
print("Second line:", second_line, end="")
````

````python
# --- Method 3: readlines() — Reads ALL lines into a LIST ---
# Each line becomes one element in the list

my_file = open("hello.txt", "r")
all_lines = my_file.readlines()     # Returns a list of strings
my_file.close()

print("Type:", type(all_lines))
print("Lines:", all_lines)

# You can loop through the list
print("\n--- Looping through lines ---")
for i, line in enumerate(all_lines, start=1):
    print(f"  Line {i}: {line.strip()}")    # .strip() removes the '\n'
````

### Writing to a File

We've already seen `write()`. Let's explore the difference between **write** (`'w'`) and **append** (`'a'`) modes.

````python
f = open("demo_write.txt", "w")
f.write("First write: this is the only content.\n")
f.close()
````

````python
f = open("demo_write.txt", "w")
f.write("Second write: the first line is gone!\n")
f.close()
````

````python
# --- Write mode ('w') vs Append mode ('a') ---

# 'w' mode: OVERWRITES the file — previous content is GONE
f = open("demo_write.txt", "w")
f.write("First write: this is the only content.\n")
f.close()

# Opening again in 'w' mode replaces everything
f = open("demo_write.txt", "w")
f.write("Second write: the first line is gone!\n")
f.close()

# Let's verify
f = open("demo_write.txt", "r")
print("After 'w' mode (overwrite):")
print(f.read())
f.close()
````

````python
# --- Append mode ('a') — Adds to the END of the file ---

# 'a' mode: keeps existing content, adds new content at the end
f = open("demo_write.txt", "a")
f.write("Third write: this was APPENDED, not overwritten.\n")
f.close()

# Let's verify — both lines should be there
f = open("demo_write.txt", "r")
print("After 'a' mode (append):")
print(f.read())
f.close()
````

````python
# --- writelines() — Write a list of strings at once ---

# Note: writelines() does NOT add newlines automatically.
# You must include '\n' yourself.

shopping_list = [
    "Eggs\n",
    "Milk\n",
    "Bread\n",
    "Butter\n"
]

f = open("shopping.txt", "w")
f.writelines(shopping_list)     # Writes all items in one call
f.close()

# Verify
f = open("shopping.txt", "r")
print(f.read())
f.close()
````

## 2. File Modes — Text vs Binary

When you open a file, Python needs to know **how** to interpret the data. There are two fundamental categories:

| Mode | Character | What It Means |
|------|-----------|---------------|
| Text | `'t'` (default) | Data is treated as human-readable strings (letters, numbers, symbols) |
| Binary | `'b'` | Data is treated as raw bytes (images, audio, executables, PDFs) |

### Complete Mode Reference

| Mode | Description | Creates file? | Overwrites? |
|------|-------------|---------------|-------------|
| `'r'` | Read (text) | No — raises error if missing | No |
| `'w'` | Write (text) | Yes | Yes |
| `'a'` | Append (text) | Yes | No — adds to end |
| `'x'` | Exclusive create (text) | Yes — raises error if exists | N/A |
| `'rb'` | Read (binary) | No | No |
| `'wb'` | Write (binary) | Yes | Yes |
| `'r+'` | Read and write | No | No (pointer at start) |
| `'w+'` | Write and read | Yes | Yes |

````python
# --- Text Mode (default) ---
# When you read a text file, Python gives you strings (str type)

f = open("hello.txt", "r")      # 'r' is the same as 'rt' (read + text)
content = f.read()
f.close()

print("Type of content:", type(content))   # <class 'str'>
print("Content:", repr(content))           # repr() shows the \n characters
````

````python
# --- Binary Mode ---
# When you read in binary mode, Python gives you bytes (bytes type)

f = open("hello.txt", "rb")     # 'rb' = read + binary
content = f.read()
f.close()

print("Type of content:", type(content))   # <class 'bytes'>
print("Content:", content)                 # Notice the b'' prefix
````

````python
# --- When do you use binary mode? ---
# Any time the file is NOT plain text: images, audio, PDFs, etc.

# Example: Copying an image file (we'll simulate with some raw bytes)
# Writing raw bytes to a binary file
sample_bytes = bytes([137, 80, 78, 71, 13, 10, 26, 10])   # PNG file header bytes

f = open("sample.bin", "wb")    # 'wb' = write + binary
f.write(sample_bytes)
f.close()

# Reading it back
f = open("sample.bin", "rb")
data = f.read()
f.close()

print("Raw bytes:", data)
print("Number of bytes:", len(data))
````

````python
# --- Exclusive Create mode ('x') ---
# Creates a new file, but raises an error if the file already exists.
# Great for preventing accidental overwrites!

import os

# Clean up first so we can demonstrate
if os.path.exists("unique_file.txt"):
    os.remove("unique_file.txt")

# First time: works fine
f = open("unique_file.txt", "x")
f.write("This file was created safely!\n")
f.close()
print("✅ File created successfully.")

# Second time: raises FileExistsError
try:
    f = open("unique_file.txt", "x")
except FileExistsError:
    print("❌ FileExistsError: The file already exists — 'x' mode won't overwrite it.")
````

## 3. Using `with` for Safe File Operations

The `with` statement (also called a **context manager**) is the **recommended** way to work with files in Python. It automatically closes the file for you — even if an error occurs.

```python
# Instead of this (risky):
f = open("file.txt", "r")
content = f.read()
f.close()

# Do this (safe):
with open("file.txt", "r") as f:
    content = f.read()
# File is automatically closed here — no need to call close()
```

````python
# --- Writing with 'with' ---

with open("with_example.txt", "w") as f:
    f.write("Line 1: The 'with' statement is elegant.\n")
    f.write("Line 2: It closes the file automatically.\n")
    f.write("Line 3: Even if an error occurs inside the block.\n")

# At this point, the file is already closed — guaranteed!
print("✅ File written and closed automatically.")
````

````python
# --- Reading with 'with' ---

with open("with_example.txt", "r") as f:
    for line_number, line in enumerate(f, start=1):   # You can loop directly over the file object!
        print(f"  Line {line_number}: {line.strip()}")

# Verify that the file is closed after the 'with' block
print(f"\nIs file closed? {f.closed}")   # True
````

````python
# --- Why 'with' is safer: error handling demo ---

# Even when an exception is raised, the file still gets closed properly.

try:
    with open("error_test.txt", "w") as f:
        f.write("Some important data\n")

        # Simulate an error
        result = 10 / 0    # ZeroDivisionError! 💥

        # This line never runs
        f.write("This won't be written\n")

except ZeroDivisionError:
    print("💥 An error occurred, but...")

# The file was still closed properly!
print(f"Is file closed? {f.closed}")   # True

# And the data written before the error was saved
with open("error_test.txt", "r") as f:
    print(f"Saved content: {f.read().strip()}")
````

````python
# --- Bonus: Working with multiple files at once ---

# You can open multiple files in a single 'with' statement
with open("hello.txt", "r") as source, open("hello_copy.txt", "w") as destination:
    content = source.read()
    destination.write(content)

print("✅ File copied successfully!")

# Verify the copy
with open("hello_copy.txt", "r") as f:
    print(f.read())
````

### Quick Reference — `open()` + `close()` vs `with`

| Feature | `open()` + `close()` | `with open()` |
|---------|---------------------|----------------|
| Must remember to close? | Yes | No (automatic) |
| Safe if error occurs? | No — file may stay open | Yes — always closes |
| Code length | Longer | Shorter |
| Recommended? | Only for learning | **Yes — always use this** |

## 4. Useful File Methods — Quick Tour

````python
# --- Checking if a file exists before opening ---

import os

filename = "hello.txt"

if os.path.exists(filename):
    print(f"✅ '{filename}' exists. Size: {os.path.getsize(filename)} bytes")
else:
    print(f"❌ '{filename}' not found.")
````

````python
# --- tell() and seek() — Controlling the read/write position ---

with open("hello.txt", "r") as f:
    # tell() shows where the "cursor" is (in bytes from the start)
    print("Starting position:", f.tell())     # 0

    # Read first 5 characters
    chunk = f.read(5)
    print(f"Read: '{chunk}'")
    print("Position after reading 5 chars:", f.tell())

    # seek() moves the cursor to a specific position
    f.seek(0)          # Go back to the beginning
    print("After seek(0):", f.tell())
    print(f"Read again: '{f.read(5)}'")
````

## 🛠️ Hands-On Activity: Log File Word Frequency Counter

Now let's put everything together! In this activity, you will:

1. Create a sample log file
2. Read and parse it to count how often each word appears
3. Write the results into a new output file

This is a very common real-world task — analyzing logs, reports, or text data.

### Step 1: Create the Sample Log File

````python
# --- Step 1: Create a realistic sample log file ---

log_content = """2024-01-15 08:23:01 INFO Server started successfully
2024-01-15 08:23:05 INFO Database connection established
2024-01-15 08:25:12 WARNING High memory usage detected
2024-01-15 08:30:00 ERROR Failed to connect to external API
2024-01-15 08:30:05 ERROR Timeout error on request to payment service
2024-01-15 08:31:00 INFO Retrying connection to external API
2024-01-15 08:31:02 INFO Connection to external API restored
2024-01-15 08:45:30 WARNING Disk usage above 80 percent
2024-01-15 09:00:00 INFO Scheduled backup started
2024-01-15 09:05:00 INFO Scheduled backup completed successfully
2024-01-15 09:15:22 ERROR Database connection lost
2024-01-15 09:15:25 INFO Attempting to reconnect to database
2024-01-15 09:15:30 INFO Database connection established
2024-01-15 09:30:00 WARNING High memory usage detected
2024-01-15 10:00:00 INFO Server health check passed
"""

# Write the log file using 'with'
with open("server.log", "w") as f:
    f.write(log_content)

print("✅ 'server.log' created with 15 log entries.")
print("\nPreview (first 3 lines):")
with open("server.log", "r") as f:
    for i, line in enumerate(f):
        if i < 3:
            print(f"  {line.strip()}")
````

### Step 2: Read the Log File and Count Word Frequencies

````python
# --- Step 2: Parse the log file and count word frequencies ---

# We'll use a dictionary to store word counts.
# Key = word (lowercase), Value = count

word_counts = {}   # Empty dictionary to hold our results

with open("server.log", "r") as f:
    for line in f:
        # Strip whitespace and skip empty lines
        line = line.strip()
        if not line:
            continue

        # Split the line into individual words
        words = line.split()

        # Count each word (converted to lowercase for consistency)
        for word in words:
            word_lower = word.lower()           # 'INFO' and 'info' count as the same word

            if word_lower in word_counts:
                word_counts[word_lower] += 1    # Word seen before — increment
            else:
                word_counts[word_lower] = 1     # First time seeing this word

# Let's see how many unique words we found
print(f"Total unique words: {len(word_counts)}")
print(f"Total word occurrences: {sum(word_counts.values())}")
word_counts
````

````python
# --- Step 2b: Sort the results by frequency (highest first) ---

# sorted() with a key function lets us sort by count (descending)
sorted_words = sorted(word_counts.items(), key=lambda pair: pair[1], reverse=True)

# Display the top 15 most frequent words
print("📊 Top 15 Most Frequent Words:")
print("-" * 35)
print(f"{'Rank':<6}{'Word':<20}{'Count'}")
print("-" * 35)

for rank, (word, count) in enumerate(sorted_words[:15], start=1):
    # Create a simple bar chart using '#' characters
    bar = "#" * count
    print(f"{rank:<6}{word:<20}{count:>3}  {bar}")
````

### Step 3: Write the Results to a New File

````python
# --- Step 3: Write the word frequency results to a new file ---

output_filename = "word_frequencies.txt"

with open(output_filename, "w") as f:
    # Write a header
    f.write("LOG FILE WORD FREQUENCY REPORT\n")
    f.write("=" * 45 + "\n")
    f.write(f"Source file: server.log\n")
    f.write(f"Total unique words: {len(word_counts)}\n")
    f.write(f"Total word occurrences: {sum(word_counts.values())}\n")
    f.write("=" * 45 + "\n\n")

    # Write the column headers
    f.write(f"{'Rank':<6}{'Word':<20}{'Count':<6}{'Bar'}\n")
    f.write("-" * 45 + "\n")

    # Write each word and its count
    for rank, (word, count) in enumerate(sorted_words, start=1):
        bar = "#" * count
        f.write(f"{rank:<6}{word:<20}{count:<6}{bar}\n")

print(f"✅ Results written to '{output_filename}'")
print(f"   File size: {os.path.getsize(output_filename)} bytes")
````

````python
# --- Step 4: Verify — Read and display the output file ---

print("📄 Contents of 'word_frequencies.txt':\n")

with open("word_frequencies.txt", "r") as f:
    print(f.read())
````

### Bonus: Count Log Levels

Let's do one more practical exercise — counting how many INFO, WARNING, and ERROR entries are in the log.

````python
# --- Bonus: Count log levels (INFO, WARNING, ERROR) ---

# We know our log format: DATE TIME LEVEL MESSAGE
# The log level is always the 3rd element (index 2) on each line.

level_counts = {}    # Dictionary to count each log level

with open("server.log", "r") as f:
    for line in f:
        line = line.strip()
        if not line:
            continue

        # Split the line and grab the log level
        parts = line.split()
        if len(parts) >= 3:
            level = parts[2]     # 'INFO', 'WARNING', or 'ERROR'
            level_counts[level] = level_counts.get(level, 0) + 1
            # .get(key, default) returns default if key doesn't exist
            # This is a cleaner alternative to the if/else pattern above

# Display results
print("📊 Log Level Summary")
print("-" * 30)
for level, count in sorted(level_counts.items()):
    bar = "█" * count
    print(f"  {level:<10} {count:>3}  {bar}")

# Write the summary to our report file (append, don't overwrite)
with open("word_frequencies.txt", "a") as f:
    f.write("\n\nLOG LEVEL SUMMARY\n")
    f.write("=" * 30 + "\n")
    for level, count in sorted(level_counts.items()):
        f.write(f"  {level:<10} {count}\n")

print("\n✅ Log level summary appended to 'word_frequencies.txt'")
````
