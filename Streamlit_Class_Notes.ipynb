## 3. Text and Display Elements

Streamlit has many ways to show text. Change the file, save, and the browser shows a **"Rerun"** button.

````python
%%writefile app2.py
import streamlit as st

st.title("Title")            # biggest text
st.header("Header")          # medium heading
st.subheader("Subheader")    # smaller heading
st.text("Plain text")        # simple fixed-width text

# Markdown lets you use **bold**, *italic* etc.
st.markdown("This is **bold** and this is *italic*")

# Coloured message boxes
st.success("Success message")
st.info("Info message")
st.warning("Warning message")
st.error("Error message")

# st.write can show almost anything: text, numbers, lists, tables
st.write(12345)
st.write(["apple", "banana", "mango"])
````

## 4. Input Widgets

Widgets let the user type, click and select.

**Important idea:** every time the user changes a widget, Streamlit **runs the whole script again from top to bottom**.

````python
%%writefile app3.py
import streamlit as st

st.title("Input Widgets")

# Text box - returns whatever the user types
name = st.text_input("What is your name?")

# Number box
age = st.number_input("What is your age?", min_value=0, max_value=100)

# Slider - (label, min, max, default)
rating = st.slider("Rate this class", 1, 10, 5)

# Dropdown
city = st.selectbox("Choose your city", ["Pune", "Mumbai", "Delhi"])

# Checkbox - returns True or False
agree = st.checkbox("I like Python")

# Button - returns True only at the moment it is clicked
if st.button("Submit"):
    st.write("Name:", name)
    st.write("Age:", age)
    st.write("Rating:", rating)
    st.write("City:", city)
    st.write("Likes Python:", agree)
````

## 5. Showing Data and Charts

````python
%%writefile app4.py
import streamlit as st
import pandas as pd

st.title("Data and Charts")

# A small table of data
data = pd.DataFrame({
    "Month": ["Jan", "Feb", "Mar", "Apr"],
    "Sales": [100, 150, 130, 200]
})

st.subheader("Our data")
st.dataframe(data)   # interactive table (you can sort columns)

# Set 'Month' as the index so charts use it on the x-axis
chart_data = data.set_index("Month")

st.subheader("Line chart")
st.line_chart(chart_data)

st.subheader("Bar chart")
st.bar_chart(chart_data)
````

## 6. Layout: Sidebar and Columns

````python
%%writefile app6.py
import streamlit as st

st.title("Layout Demo")

# Anything with st.sidebar goes in the left panel
st.sidebar.header("Settings")
color = st.sidebar.selectbox("Pick a colour", ["Red", "Green", "Blue"])

st.write("You picked:", color)

# Split the page into 2 columns side by side
col1, col2 = st.columns(2)

with col1:
    st.header("Left")
    st.write("This is column 1")

with col2:
    st.header("Right")
    st.write("This is column 2")
````

## 7. Mini Project: BMI Calculator

Putting it all together.

BMI = weight (kg) / (height in metres)²

````python
%%writefile bmi_app.py
import streamlit as st

st.title("BMI Calculator")

weight = st.number_input("Weight (kg)", min_value=1.0, value=60.0)
height = st.number_input("Height (cm)", min_value=50.0, value=170.0)

if st.button("Calculate"):
    height_m = height / 100           # convert cm to metres
    bmi = weight / (height_m ** 2)    # BMI formula

    st.write("Your BMI is:", round(bmi, 1))

    # Show a different message based on the result
    if bmi < 18.5:
        st.warning("Underweight")
    elif bmi < 25:
        st.success("Normal")
    else:
        st.error("Overweight")
````

## 8. FastAPI Basics

**Streamlit** = the front end (what the user sees)
**FastAPI** = the back end (logic that sends back data)

An **API** is a URL that returns data (usually JSON) instead of a web page.

````python
%%writefile api.py
from fastapi import FastAPI
import uvicorn   # NEW: needed to start the API from the code

app = FastAPI()   # create the API


# Visiting http://localhost:8001/ runs this function
@app.get("/")
def home():
    return {"message": "Hello from FastAPI"}


# {name} is taken from the URL. Example: /hello/Rahul
@app.get("/hello/{name}")
def say_hello(name: str):
    return {"message": "Hello " + name}


# a and b come from the URL. Example: /add?a=5&b=3
@app.get("/add")
def add(a: int, b: int):
    return {"result": a + b}


# Same BMI logic, now inside the API
@app.get("/bmi")
def bmi(weight: float, height: float):
    height_m = height / 100
    value = weight / (height_m ** 2)
    return {"bmi": round(value, 1)}


# NEW: start the API on port 8001
if __name__ == "__main__":
    uvicorn.run(app, port=8001)
````

### How to run the API
In CMD, from the folder where the file is:  
   python api.py  
In a **terminal**:

```
uvicorn api:app --reload
```

- `api` = the file name (`api.py`)
- `app` = the variable name inside the file
- `--reload` = restart automatically when you save changes

Now try these in your browser:
- http://localhost:8001/
- http://localhost:8001/hello/Rahul
- http://localhost:8001/add?a=5&b=3
- http://localhost:8001/docs  ← FastAPI gives you free documentation where you can test every endpoint!

### Calling the API from Python

With the API running in the terminal, run this cell:

````python
import requests

# Send a GET request to our API
response = requests.get("http://localhost:8001/add", params={"a": 10, "b": 20})

# .json() converts the reply into a Python dictionary
print(response.json())
````

## 9. Connecting Streamlit + FastAPI

The Streamlit app takes input from the user, sends it to FastAPI, and shows the answer.

You need **two terminals**:  
1. Close all CMD windows.
2. Open CMD window 1 and start the API:
    cd Desktop  
    python api.py

   Now look carefully at what it prints:  
   Uvicorn running on http://127.0.0.1:8001: the API is working. Leave this window open and go to step 3.  
3. Test in the browser: open http://localhost:8001/. You should see {"message":"Hello from FastAPI"}.
4. Open CMD window 2 and start Streamlit:
   cd Desktop  
   streamlit run frontend.py  



````python
%%writefile frontend.py
import streamlit as st
import requests

API_URL = "http://localhost:8001"   # where FastAPI is running

st.title("BMI Calculator (with FastAPI)")

weight = st.number_input("Weight (kg)", min_value=1.0, value=60.0)
height = st.number_input("Height (cm)", min_value=50.0, value=170.0)

if st.button("Calculate"):
    # Ask the API to do the calculation
    response = requests.get(API_URL + "/bmi", params={"weight": weight, "height": height})
    data = response.json()

    st.success("Your BMI is: " + str(data["bmi"]))
````
