## Welcome!

Object-Oriented Programming is a way of organizing code around **objects** — bundles of data and the actions that go with that data.

## 1. Classes and Objects

### 1.1 What is a Class?

A **class** is a template that defines the structure and behaviour that its objects will have.  
An **object** (also called an **instance**) is a concrete thing created from that template.

**Real-world analogy:**  
A *cookie cutter* (class) isn't a cookie — but every cookie (object) you stamp out with it has the same shape.

````python
# ── Our very first class ──────────────────────────────────────────────
# The simplest possible class — it does nothing yet, but it's valid.

class Dog:
    pass                         # 'pass' is a placeholder; the body is empty

# Creating objects (instances) from the class
dog1 = Dog()                     # dog1 is an instance of Dog
dog2 = Dog()                     # dog2 is a separate instance

# Each object lives at a different memory address
print(type(dog1))                # <class '__main__.Dog'>
print(dog1 is dog2)              # False — two distinct objects
print(id(dog1))
print(id(dog2))
````

### 1.2 Why Use Classes?

Without classes, related data lives in separate variables that can easily get out of sync:

```python
# Fragile — nothing ties these together
dog1_name = "Buddy"
dog1_breed = "Labrador"
dog1_age = 3
```

With a class, name, breed, and age travel together as a single object — impossible to lose or mix up.

## 2. Attributes, Methods, and `__init__`

### 2.1 The `__init__` Method (Constructor)

`__init__` is a special method that Python calls **automatically** every time you create a new object. It sets up the object's initial state.

| Term | Meaning |
|------|---------|
| `self` | A reference to the *current* object being created or used |
| `__init__` | The initialiser — runs once at object creation |
| Attribute | A variable that belongs to an object (e.g. `self.name`) |
| Method | A function that belongs to a class (e.g. `def bark(self)`) |

````python
# ── A class with __init__, attributes, and methods ───────────────────

class Dog:
    # Class attribute — shared by ALL Dog instances
    species = "Canis familiaris"

    def __init__(self, name, breed, age):
        """Initialise a new Dog object.

        Parameters
        ----------
        name  : str   — the dog's name
        breed : str   — the dog's breed
        age   : int   — the dog's age in years
        """
        # Instance attributes — unique to each object
        self.name = name             # store name on this specific dog
        self.breed = breed
        self.age = age

    # Instance method — operates on a single object
    def bark(self):
        """Make the dog bark."""
        return f"{self.name} says: Woof!"

    def description(self):
        """Return a human-readable summary."""
        return f"{self.name} is a {self.age}-year-old {self.breed}."


# ── Creating and using objects ───────────────────────────────────────
buddy = Dog("Buddy", "Labrador", 3)
max_dog = Dog("Max", "Beagle", 5)

print(buddy.bark())              # Buddy says: Woof!
print(max_dog.description())     # Max is a 5-year-old Beagle.
print(buddy.species)             # Canis familiaris (class attribute)
````

### 2.2 Class Attributes vs Instance Attributes

| Feature | Class Attribute | Instance Attribute |
|---------|----------------|-------------------|
| Defined where? | Directly in the class body | Inside `__init__` using `self.` |
| Shared? | Yes — same value for every instance | No — each instance has its own copy |
| Access | `ClassName.attr` or `self.attr` | `self.attr` only |
| Example | `species = "Canis familiaris"` | `self.name = name` |

````python
# ── Class vs instance attributes in action ───────────────────────────

class Counter:
    total_counters = 0              # class attribute — tracks ALL counters

    def __init__(self, label):
        self.label = label          # instance attribute — unique per counter
        self.count = 0              # instance attribute
        Counter.total_counters += 1 # modify the shared class attribute

    def increment(self):
        self.count += 1

# Create two counters
c1 = Counter("Page Views")
c2 = Counter("Button Clicks")

c1.increment()
c1.increment()
c2.increment()

print(f"{c1.label}: {c1.count}")       # Page Views: 2
print(f"{c2.label}: {c2.count}")       # Button Clicks: 1
print(f"Total counters: {Counter.total_counters}")  # Total counters: 2
````

### ⚠️ Common Pitfall: Forgetting `self`

````python
# ── What happens when you forget self? ────────────────────────────────

class BrokenDog:
    def __init__(self, name):
        name = name        # ❌ This creates a LOCAL variable, not an attribute!

broken = BrokenDog("Buddy")

try:
    print(broken.name)     # ❌ AttributeError — 'name' was never stored on the object
except AttributeError as e:
    print(f"Error: {e}")
    print("Fix: use 'self.name = name' inside __init__")
````

## 3. Inheritance

Inheritance lets you create a **child class** that automatically gets all the attributes and methods of a **parent class**, and then add or override whatever is different.

**Real-world analogy:**  
A *smartphone* inherits everything a basic phone can do (make calls, send texts) and adds new capabilities (apps, camera, GPS).

```
        Animal          ← parent (base) class
       /      \
     Dog      Cat       ← child (derived) classes
```

````python
 # ── Basic inheritance ─────────────────────────────────────────────────

class Animal:
    """Parent class for all animals."""

    def __init__(self, name, sound):
        self.name = name
        self.sound = sound

    def speak(self):
        return f"{self.name} says {self.sound}!"

    # Defining the human-readable string representation
    def __str__(self):
       return f"Animal(name='{self.name}')"


class Dog(Animal):
    """Dog inherits from Animal and adds tricks."""

    def __init__(self, name, breed):
        # Call the parent's __init__ to set name and sound
        super().__init__(name, sound="Woof")
        self.breed = breed          # new attribute specific to Dog
        self.tricks = []            # new attribute specific to Dog

    def learn_trick(self, trick):
        """Teach the dog a new trick."""
        self.tricks.append(trick)

    def show_tricks(self):
        if self.tricks:
            return f"{self.name} knows: {', '.join(self.tricks)}"
        return f"{self.name} hasn't learned any tricks yet."


class Cat(Animal):
    """Cat inherits from Animal and adds purring."""

    def __init__(self, name, indoor=True):
        super().__init__(name, sound="Meow")
        self.indoor = indoor

    def purr(self):
        return f"{self.name} purrs softly... 🐱"


# ── Using inherited and new methods ──────────────────────────────────
buddy = Dog("Buddy", "Golden Retriever")
whiskers = Cat("Whiskers", indoor=True)

# Printing the object directly triggers __str__()
print(f"{buddy}")

# speak() is inherited from Animal — works for both
print(buddy.speak())             # Buddy says Woof!
print(whiskers.speak())          # Whiskers says Meow!

# Dog-specific methods
buddy.learn_trick("sit")
buddy.learn_trick("shake")
print(buddy.show_tricks())       # Buddy knows: sit, shake

# Cat-specific method
print(whiskers.purr())           # Whiskers purrs softly... 🐱
````

### 3.1 The `super()` Function

`super()` gives you access to the **parent class's methods**. The most common use is calling the parent's `__init__` so you don't have to duplicate its setup code.

```python
# Without super() — you'd have to repeat the parent's work
class Dog(Animal):
    def __init__(self, name, breed):
        self.name = name           # ❌ duplicating Animal's job
        self.sound = "Woof"        # ❌ duplicating Animal's job
        self.breed = breed

# With super() — clean and DRY (Don't Repeat Yourself)
class Dog(Animal):
    def __init__(self, name, breed):
        super().__init__(name, sound="Woof")  # ✅ let Animal handle it
        self.breed = breed
```

### 3.2 Checking Inheritance Relationships

````python
# ── isinstance() and issubclass() ─────────────────────────────────────

buddy = Dog("Buddy", "Labrador")

# isinstance checks if an object is an instance of a class (or its parents)
print(isinstance(buddy, Dog))      # True  — buddy IS a Dog
print(isinstance(buddy, Animal))   # True  — a Dog IS also an Animal
print(isinstance(buddy, Cat))      # False — a Dog is NOT a Cat

# issubclass checks the class hierarchy itself
print(issubclass(Dog, Animal))     # True  — Dog inherits from Animal
print(issubclass(Cat, Animal))     # True
print(issubclass(Dog, Cat))        # False

# The Method Resolution Order (MRO) shows the lookup chain
print(Dog.__mro__)
# (<class 'Dog'>, <class 'Animal'>, <class 'object'>)
````

## 4. Polymorphism

**Polymorphism** means "many forms." In practice, it means you can call the **same method name** on different objects and each object responds in its own way.

**Real-world analogy:**  
The command "Start!" means different things to a runner (sprint), a car (turn the engine on), and a computer (boot up) — same word, different actions.

````python
# ── Polymorphism through method overriding ────────────────────────────

class Shape:
    """Base class for geometric shapes."""

    def area(self):
        raise NotImplementedError("Subclasses must implement area()")

    def describe(self):
        return f"I am a {type(self).__name__} with area {self.area():.2f}"


class Rectangle(Shape):
    def __init__(self, width, height):
        self.width = width
        self.height = height

    def area(self):
        return self.width * self.height           # Rectangle's version


class Circle(Shape):
    def __init__(self, radius):
        self.radius = radius

    def area(self):
        import math
        return math.pi * self.radius ** 2         # Circle's version


class Triangle(Shape):
    def __init__(self, base, height):
        self.base = base
        self.height = height

    def area(self):
        return 0.5 * self.base * self.height      # Triangle's version


# ── Same interface, different behaviour ──────────────────────────────
shapes = [
    Rectangle(5, 3),
    Circle(4),
    Triangle(6, 8),
]

# We can loop over all shapes and call .area() without caring which type
for shape in shapes:
    print(shape.describe())
# I am a Rectangle with area 15.00
# I am a Circle with area 50.27
# I am a Triangle with area 24.00
````

### Why Polymorphism Matters

The `for` loop above doesn't need `if/elif` checks for each shape type. When you add a new shape (e.g. `Pentagon`), the loop works *without any changes* — you only write the new class. This is the **Open/Closed Principle**: open for extension, closed for modification.

## 📋 Quick Reference

### OOP Pillars at a Glance

| Pillar | What it does | Python mechanism |
|--------|-------------|-----------------|
| **Abstraction** | Hide complexity, show only what matters | Classes with clean public methods |
| **Encapsulation** | Bundle data + methods; control access | `_protected`, `__private`, `@property` |
| **Inheritance** | Create child classes from parent classes | `class Child(Parent):`, `super()` |
| **Polymorphism** | Same method name, different behaviour | Method overriding in subclasses |

Create a `Student` class with attributes for `name`, `student_id`, and `grades` (a list). Add methods to add a grade, calculate the average, and display the student's info.

````python
# ── Assignment 1: Student Record ──────────────────────────────────────

class Student:
    def __init__(self, name, student_id):
        # TODO: Store name, student_id, and an empty list for grades
        self.name = name
        self.student_id = student_id
        self.grades = []

    def add_grade(self, grade):
        # TODO: Append grade to the grades list (validate 0-100)
        if 0 <= int(grade) <= 100:
          self.grades.append(grade)

    def average(self):
        # TODO: Return the average of all grades, or 0 if no grades
        if len(self.grades) != 0:
          return (f"{sum(self.grades)/len(self.grades)}")
        else:
          return 0

    def __str__(self):
        # TODO: Return a string like "Alice (ID: 101) — Avg: 85.5"
        avg = self.average()
        return f"{self.name} (ID: {self.student_id}) '-' Avg: {avg}"


# ── Test your solution ───────────────────────────────────────────────
s = Student("Alice", 101)
s.add_grade(90)
s.add_grade(85)
s.add_grade(92)
print(s)                    # Alice (ID: 101) — Avg: 89.0
#print(s.average())          # 89.0
````
