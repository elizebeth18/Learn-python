## 1. Lambda Functions

### 1.1 What is a Lambda?

A **lambda** is a small, anonymous (unnamed) function defined in a single expression. It can take any number of arguments but contains only **one expression** — and it automatically returns the result.

| Feature | Regular Function (`def`) | Lambda Function |
|---------|------------------------|-----------------|
| Syntax | `def name(args): return ...` | `lambda args: expression` |
| Name | Has a name | Anonymous (no name) |
| Body | Multiple statements allowed | Single expression only |
| Use case | Reusable, complex logic | Quick, one-off operations |

````python
# ── Your first lambda ─────────────────────────────────────────────────

# Regular function
def square(x):
    return x ** 2

# Equivalent lambda
square_lambda = lambda x: x ** 2      # lambda arguments: expression

# Both produce the same result
print(square(5))                       # 25
print(square_lambda(5))                # 25

# But notice the type
print(type(square))                    # <class 'function'>
print(type(square_lambda))            # <class 'function'> — same type!
````

### 1.2 Lambda Syntax Breakdown

```
lambda  x, y  :  x + y
  │      │         │
  │      │         └── expression (automatically returned)
  │      └──────────── arguments (comma-separated, no parentheses needed)
  └─────────────────── the lambda keyword
```

> **Key rule:** A lambda can only contain a **single expression** — no `if` statements, no loops, no assignments. However, it *can* use a conditional expression (ternary operator).

````python
# ── Lambdas with different numbers of arguments ──────────────────────

# No arguments
greet = lambda: "Hello, World!"
print(greet())                         # Hello, World!

# One argument
double = lambda x: x * 2
print(double(7))                       # 14

# Two arguments
add = lambda x, y: x + y
print(add(3, 5))                       # 8

# Multiple arguments with a calculation
bmi = lambda weight_kg, height_m: round(weight_kg / height_m ** 2, 1)
print(bmi(70, 1.75))                   # 22.9

# Using a conditional expression (ternary) inside a lambda
check_even = lambda n: "Even" if n % 2 == 0 else "Odd"
print(check_even(4))                   # Even
print(check_even(7))                   # Odd
````

## 2. The `map()` Function

### 2.1 What Does `map()` Do?

`map()` applies a function to **every item** in an iterable (list, tuple, etc.) and returns a new iterable with the results.

```
map(function, iterable)
  │      │         │
  │      │         └── the data to process (list, tuple, string, etc.)
  │      └──────────── what to do to each item
  └─────────────────── the map function itself
```

> **Important:** `map()` returns a **map object** (a lazy iterator), not a list. Wrap it in `list()` to see all the results at once.

````python
# ── map() with a regular function ─────────────────────────────────────

def to_uppercase(text):
    """Convert a string to uppercase."""
    return text.upper()

# Without map — manual loop
names = ["alice", "bob", "charlie"]
result = []
for name in names:
    result.append(to_uppercase(name))
print(result)                          # ['ALICE', 'BOB', 'CHARLIE']

# With map — one clean line
result = list(map(to_uppercase, names))
print(result)                          # ['ALICE', 'BOB', 'CHARLIE']
````

````python
# ── What does map() actually return? ──────────────────────────────────

numbers = [1, 2, 3, 4, 5]
result = map(lambda x: x ** 2, numbers)

print(result)                          # <map object at 0x...> — NOT a list!
print(type(result))                    # <class 'map'>

# Convert to list to see the values
print(list(result))                    # [1, 4, 9, 16, 25]

# ⚠️ Careful: map objects are exhausted after one pass
result = map(lambda x: x ** 2, numbers)
print(list(result))                    # [1, 4, 9, 16, 25] — first time works
print(list(result))                    # [] — empty! already consumed
````

### 2.2 `map()` with Built-in Functions

You don't always need lambda or custom functions — Python's built-in functions work perfectly with `map()`.

````python
# ── map() with built-in functions ─────────────────────────────────────

# Convert strings to integers
str_numbers = ["10", "20", "30", "40"]
int_numbers = list(map(int, str_numbers))
print(int_numbers)                     # [10, 20, 30, 40]

# Get the length of each word
words = ["Python", "is", "awesome"]
lengths = list(map(len, words))
print(lengths)                         # [6, 2, 7]

# Convert to floats
prices = ["9.99", "24.50", "3.75"]
float_prices = list(map(float, prices))
print(float_prices)                    # [9.99, 24.5, 3.75]

# Absolute values
temps = [-5, 12, -3, 8, -1]
abs_temps = list(map(abs, temps))
print(abs_temps)                       # [5, 12, 3, 8, 1]
````

### 2.3 `map()` with Multiple Iterables

`map()` can take **more than one iterable**. The function must then accept that many arguments, and `map()` pulls one item from each iterable per call.

````python
# ── map() with two iterables ──────────────────────────────────────────

# Add corresponding elements from two lists
list_a = [1, 2, 3, 4]
list_b = [10, 20, 30, 40]

sums = list(map(lambda x, y: x + y, list_a, list_b))
print(sums)                            # [11, 22, 33, 44]

# Combine first and last names
first_names = ["Alice", "Bob", "Charlie"]
last_names = ["Smith", "Jones", "Brown"]

full_names = list(map(lambda f, l: f"{f} {l}", first_names, last_names))
print(full_names)                      # ['Alice Smith', 'Bob Jones', 'Charlie Brown']

# ⚠️ If iterables have different lengths, map stops at the shortest
a = [1, 2, 3, 4, 5]
b = [10, 20, 30]
print(list(map(lambda x, y: x + y, a, b)))  # [11, 22, 33] — stops at 3 items
````

## 3. Lambda + Map Together

This is where the magic happens. Combining `lambda` with `map()` gives you concise, readable one-liners for data transformation.

````python
# ── Practical examples: lambda + map ──────────────────────────────────

# Example 1: Convert temperatures from Celsius to Fahrenheit
celsius = [0, 20, 37, 100]
fahrenheit = list(map(lambda c: (c * 9/5) + 32, celsius))
print(f"Celsius:    {celsius}")        # [0, 20, 37, 100]
print(f"Fahrenheit: {fahrenheit}")     # [32.0, 68.0, 98.6, 212.0]
````

````python
# Example 2: Extract specific data from a list of dictionaries
students = [
    {"name": "Alice", "grade": 92},
    {"name": "Bob", "grade": 78},
    {"name": "Charlie", "grade": 85},
]

# Pull out just the names
names = list(map(lambda s: s["name"], students))
print(names)                           # ['Alice', 'Bob', 'Charlie']

# Format grade reports
reports = list(map(
    lambda s: f"{s['name']}: {'Pass' if s['grade'] >= 80 else 'Needs Improvement'}",
    students
))
for r in reports:
    print(r)
# Alice: Pass
# Bob: Needs Improvement
# Charlie: Pass
````

````python
# Example 3: Clean and normalise user input
raw_inputs = ["  Alice  ", "BOB", "  charlie ", "  DIANA"]

# Strip whitespace and capitalise properly — all in one step
cleaned = list(map(lambda s: s.strip().title(), raw_inputs))
print(cleaned)                         # ['Alice', 'Bob', 'Charlie', 'Diana']
````

````python
# Example 4: Apply a discount to a list of prices
prices = [49.99, 29.99, 99.99, 14.99]

# 20% discount
discounted = list(map(lambda p: round(p * 0.80, 2), prices))
print(f"Original:   {prices}")         # [49.99, 29.99, 99.99, 14.99]
print(f"Discounted: {discounted}")     # [39.99, 24.0, 80.0, 11.99]
````

## 📋 Quick Reference

### Lambda Syntax

```python
# Basic form
lambda arguments: expression

# Examples
lambda x: x + 1                       # one argument
lambda x, y: x * y                    # two arguments
lambda: 42                            # no arguments
lambda s: s.strip().lower()           # chained methods
lambda n: "Even" if n % 2 == 0 else "Odd"   # conditional expression
```

### map(), filter(), sorted() with Lambda

```python
# map — transform every item
list(map(lambda x: x * 2, [1, 2, 3]))               # [2, 4, 6]

# filter — keep items where lambda returns True
list(filter(lambda x: x > 2, [1, 2, 3, 4]))          # [3, 4]

# sorted — custom sort order
sorted(["cat", "ab", "dogs"], key=lambda s: len(s))   # ['ab', 'cat', 'dogs']

# Combining map and filter
list(map(lambda x: x**2, filter(lambda x: x > 0, [-2, -1, 0, 1, 2])))  # [1, 4]
```
