# Python Functions

## 1. Defining and Calling Functions

A function is defined using the `def` keyword, followed by a name, parentheses, and a colon. The indented block below it is the function **body** — the code that runs when the function is called.

```python
def function_name(parameters):
    """Optional docstring — describes what the function does."""
    # function body
    return result    # optional
```

````python
def my_sum_function(num1, num2):
    """Add two numbers and return the result."""
    result = num1 + num2
    return result
````

````python
my_sum_function(15, 20)
````

````python
# --- Your first function ---
# This function takes no parameters and simply prints a greeting.

def say_hello():
    """Print a friendly greeting."""
    print("Hello! Welcome to Python functions.")

# IMPORTANT: Defining a function does NOT run it.
# You must CALL it by writing its name followed by parentheses.

say_hello()     # This is the function call — now the code inside runs
````

````python
# --- A function with one parameter ---
# Parameters are variables that receive values when the function is called.

def greet(name):
    """Greet a person by their name."""
    print(f"Hello, {name}! Nice to meet you.")

# Call it with different arguments
greet("Alice")
greet("Bob")
greet("Charlie")
````

````python
# --- A function with multiple parameters ---

def add_numbers(a, b):
    """Add two numbers and print the result."""
    result = a + b
    print(f"{a} + {b} = {result}")

add_numbers(3, 5)
add_numbers(100, 250)
add_numbers(-10, 10)
````

````python
# --- Calling a function multiple times shows the power of reuse ---
# Without functions, you'd copy-paste this logic every time.

def draw_box(text):
    """Draw a simple text box around a message."""
    border = "+" + "-" * (len(text) + 2) + "+"
    print(border)
    print(f"| {text} |")
    print(border)

draw_box("Hello")
print()                     # Blank line between boxes
draw_box("Python is fun!")
print()
draw_box("Functions save time")
````

### Key Terminology

| Term | Meaning | Example |
|------|---------|--------|
| **Parameter** | Variable name in the function definition | `name` in `def greet(name):` |
| **Argument** | Actual value passed when calling the function | `"Alice"` in `greet("Alice")` |
| **Call / Invoke** | Running the function by using its name with `()` | `greet("Alice")` |

## 2. Arguments — Positional, Keyword, Default, *args, and **kwargs

Python gives you many flexible ways to pass data into functions. Let's explore each one.

### 2.1 Positional Arguments

The most basic type — arguments are matched to parameters **by their position** (left to right).

````python
# --- Positional arguments: ORDER MATTERS ---

def describe_pet(animal, name):
    """Display information about a pet."""
    print(f"I have a {animal} named {name}.")

# Correct order: first argument → animal, second → name
describe_pet("dog", "Buddy")

# Swapped order: the meaning changes!
describe_pet("Buddy", "dog")
# ↑ This says "I have a Buddy named dog" — not what we want!
````

````python
# --- What happens if you pass too few or too many arguments? ---

def introduce(first_name, last_name):
    print(f"My name is {first_name} {last_name}.")

# Correct: exactly 2 arguments for 2 parameters
introduce("John", "Doe")

# Uncomment the lines below to see the errors:
# introduce("John")              # TypeError: missing 1 required argument
# introduce("John", "Doe", 25)   # TypeError: takes 2 arguments but 3 were given
````

### 2.2 Keyword Arguments

You can pass arguments by **name**, which makes your code clearer and lets you put them in any order.

````python
# --- Keyword arguments: use parameter_name=value ---

def describe_pet(animal, name):
    """Display information about a pet."""
    print(f"I have a {animal} named {name}.")

# Using keyword arguments — order doesn't matter anymore!
describe_pet(animal="cat", name="Whiskers")
describe_pet(name="Goldie", animal="fish")     # Swapped, but still correct

# You can also mix positional and keyword arguments,
# but positional must come FIRST
describe_pet("hamster", name="Nibbles")         # animal is positional, name is keyword
describe_pet("star fish", "fish")
# describe_pet(animal = "hamster", "Nibbles") ### this will throw error (SyntaxError: positional argument follows keyword argument)
````

````python
# --- Why keyword arguments are useful: readability ---
# Which call is easier to understand at a glance?

def create_user(username, age, country, is_active):
    """Simulate creating a user account."""
    status = "Active" if is_active else "Inactive"
    print(f"User: {username}, Age: {age}, Country: {country}, Status: {status}")

# Positional — you have to remember the order
create_user("alice99", 28, "Canada", True)

# Keyword — self-documenting and much clearer
create_user(username="bob_42", age=35, country="Germany", is_active=False)
````

### 2.3 Default Arguments

You can give parameters a **default value**. If the caller doesn't provide that argument, the default is used.

````python
# --- Default arguments: provide fallback values ---

def greet(name, greeting="Hello"):
    """Greet someone with a customizable greeting."""
    print(f"{greeting}, {name}!")

# Using the default greeting
greet("Alice")                          # Uses default: "Hello"

# Overriding the default
greet("Bob", greeting="Good morning")   # Custom greeting
greet("Charlie", "Hey there")           # Also works positionally
````

````python
# --- Practical example: a configurable function ---

def calculate_total(price, tax_rate=0.08, discount=0):
    """
    Calculate the total price after tax and discount.

    Parameters:
        price     : Base price of the item
        tax_rate  : Tax rate as a decimal (default 8%)
        discount  : Discount amount in dollars (default $0)
    """
    subtotal = price - discount
    tax = subtotal * tax_rate
    total = subtotal + tax
    print(f"Price: ${price:.2f} | Discount: ${discount:.2f} | Tax: ${tax:.2f} | Total: ${total:.2f}")

# Use all defaults (8% tax, no discount)
calculate_total(100)

# Custom tax rate, default discount
calculate_total(100, tax_rate=0.10)

# Custom discount, default tax
calculate_total(100, discount=15)

# Override everything
calculate_total(100, tax_rate=0.05, discount=20)
````

````python
# --- ⚠️ Rule: Default parameters must come AFTER non-default ones ---

# ✅ Correct
def example_good(name, age=25):
    print(f"{name} is {age}")

example_good("Alice")

# ❌ Incorrect — uncomment to see the SyntaxError
# def example_bad(age=25, name):
#     print(f"{name} is {age}")
# SyntaxError: non-default argument follows default argument
````

### 2.4 `*args` — Variable Number of Positional Arguments

Sometimes you don't know in advance how many arguments someone will pass. The `*args` syntax collects any number of positional arguments into a **tuple**.

````python
# --- *args collects extra positional arguments into a tuple ---

def add_all(*args):
    """Add up any number of values."""
    print(f"Received: {args}")          # args is a tuple
    print(f"Type:     {type(args)}")
    print(f"Sum:      {sum(args)}")
    print()

# Call with different numbers of arguments — all work!
add_all(1, 2)
add_all(10, 20, 30)
add_all(5, 10, 15, 20, 25)
````

````python
# --- Combining regular parameters with *args ---
# Regular parameters are matched first; the rest go into *args.

def introduce_team(team_name, *members):
    """Introduce a team and list its members."""
    print(f"Team: {team_name}")
    print(f"Members ({len(members)}):")
    for i, member in enumerate(members, start=1):
        print(f"  {i}. {member}")
    print()

introduce_team("Alpha", "Alice", "Bob", "Charlie")
introduce_team("Beta", "Diana", "Eve")
````

````python
# --- Looping through *args ---

def find_maximum(*args): # ex: 3 , 2 , 5
    """Find the largest number among the arguments."""
    if not args:                        # Handle the case of no arguments
        print("No numbers provided!")
        return

    largest = args[0]                   # Start with the first value
    for num in args[1:]:                # Compare with the rest
        if num > largest:
            largest = num

    print(f"Numbers: {args}  →  Maximum: {largest}")

find_maximum(3, 7, 2, 9, 1)
find_maximum(42)
find_maximum(-5, -1, -10)
````

### 2.5 `**kwargs` — Variable Number of Keyword Arguments

Just like `*args` collects extra positional arguments, `**kwargs` collects extra **keyword arguments** into a **dictionary**.

````python
# --- **kwargs collects extra keyword arguments into a dictionary ---

def print_info(**kwargs):
    """Print any information passed as keyword arguments."""
    print(f"Received: {kwargs}")        # kwargs is a dictionary
    print(f"Type:     {type(kwargs)}")
    print("Details:")
    for key, value in kwargs.items():
        print(f"  {key}: {value}")
    print()

# Pass any key=value pairs you want
print_info(name="Alice", age=30, city="New York")
print_info(language="Python", level="beginner")
my_data = {"name": "Bob", "age": 25, "city": "London"}
print_info(**my_data)
````

````python
# --- Practical example: building a profile card ---

def build_profile(first_name, last_name, **details):
    """
    Build a user profile. First and last name are required;
    any additional details are optional.
    """
    profile = {
        "first_name": first_name,
        "last_name": last_name,
    }
    # Merge the extra keyword arguments into the profile
    profile.update(details)
    return profile

# Minimal profile
user1 = build_profile("Alice", "Smith")
print("User 1:", user1)

# Rich profile with extra details
user2 = build_profile("Bob", "Johnson", age=28, city="London", hobby="chess")
print("User 2:", user2)
````

````python
# --- Combining *args and **kwargs ---
# You can use both in the same function!
# Order must be: regular params → *args → **kwargs

def log_event(event_type, *tags, **details):
    """
    Log an event with optional tags and details.

    Parameters:
        event_type : str           (positional — required)
        *tags      : str           (extra positional — optional tags)
        **details  : key=value     (extra keyword — optional metadata)
    """
    print(f"Event: {event_type}")

    if tags:
        print(f"  Tags: {', '.join(tags)}")

    if details:
        for key, value in details.items():
            print(f"  {key}: {value}")
    print()

# Different ways to call it
log_event("LOGIN")
log_event("PURCHASE", "electronics", "sale", amount=299.99, currency="USD")
log_event("ERROR", "critical", "database", code=500, message="Connection lost")
````

### Argument Order Summary

When defining a function, parameters must follow this order:

```python
def func(positional, keyword=default, *args, **kwargs):
    ...
```

| # | Type | Example | Collected as |
|---|------|---------|-------------|
| 1 | Regular positional | `name` | single value |
| 2 | Default value | `age=25` | single value |
| 3 | `*args` | `*args` | tuple |
| 4 | `**kwargs` | `**kwargs` | dictionary |

## 3. Scope and Return Values

### 3.1 Return Values

So far most of our functions have used `print()` to show output. But in real programs, you usually want a function to **give back** a value so you can use it elsewhere. That's what `return` does.

````python
# --- print() vs return: a critical distinction ---

# This function PRINTS the result (shows it on screen, but doesn't give it back)
def add_print(a, b):
    print(a + b)

# This function RETURNS the result (gives it back so you can use it)
def add_return(a, b):
    return a + b

# Let's see the difference
result1 = add_print(3, 5)      # Prints 8 to screen
result2 = add_return(3, 5)     # Silently returns 8

print(f"result1 (from print version): {result1}")   # None! print() doesn't give a value back
print(f"result2 (from return version): {result2}")   # 8 — we can use this!

# With return, you can chain operations
total = add_return(3, 5) + add_return(10, 20)
print(f"Chained result: {total}")    # 38
````

````python
# --- Returning different types ---

# Return a string
def format_name(first, last):
    """Return a formatted full name."""
    return f"{first.title()} {last.title()}"

full_name = format_name("john", "doe")
print(full_name)     # John Doe

# Return a boolean
def is_adult(age):
    """Check if someone is 18 or older."""
    return age >= 18

print(is_adult(21))    # True
print(is_adult(15))    # False

# Use it in a condition
if is_adult(21):
    print("Access granted.")
````

````python
# --- Returning multiple values ---
# Python lets you return multiple values as a tuple.

def get_min_max(numbers):
    """Return both the minimum and maximum of a list."""
    return min(numbers), max(numbers)       # Returns a tuple: (min, max)

# You can unpack the returned tuple into separate variables
data = [45, 12, 78, 3, 91, 56]

smallest, largest = get_min_max(data)       # Tuple unpacking
print(f"Data: {data}")
print(f"Data: {get_min_max(data)}")
print(f"Smallest: {smallest}")
print(f"Largest:  {largest}")
````

````python
# --- Returning a dictionary for named results ---

def analyze_text(text):
    """Analyze a text string and return statistics."""
    words = text.split()
    return {
        "characters": len(text),
        "words": len(words),
        "sentences": text.count(".") + text.count("!") + text.count("?"),
        "avg_word_length": round(sum(len(w) for w in words) / len(words), 1)
    }

sample = "Python is great. Functions make code reusable. Keep learning!"
stats = analyze_text(sample)

print(f"Text: \"{sample}\"")
print(stats.items())
for key, value in stats.items():
    print(f"  {key}: {value}")
````

````python
# --- Early return: exiting a function before the end ---
# 'return' immediately stops the function, even inside a loop.

def find_first_negative(numbers):
    """Return the first negative number in a list, or None if there are none."""
    for num in numbers:
        if num < 0:
            return num          # Found one — stop immediately
    return None                 # Reached the end with no negatives

print(find_first_negative([3, 7, -2, 5, -8]))    # -2 (stops at the first one)
print(find_first_negative([1, 2, 3, 4]))          # None (no negatives found)
````

### 3.2 Scope — Where Variables Live

**Scope** determines where a variable can be accessed. Python has two main scopes:

| Scope | Where | Accessible from |
|-------|-------|----------------|
| **Local** | Inside a function | Only within that function |
| **Global** | Outside all functions | Anywhere in the file |

````python
# --- Local scope: variables created inside a function stay inside ---

def my_function():
    secret = "I only exist inside this function"    # Local variable
    print(f"Inside the function: {secret}")

my_function()

# Trying to access 'secret' outside the function causes an error
try:
    print(secret)
except NameError as e:
    print(f"Outside the function: NameError — {e}")
````

````python
# --- Global scope: variables created outside are visible inside functions ---

greeting = "Hello"              # Global variable

def say_greeting():
    # Functions can READ global variables
    print(f"{greeting}, World!")    # Uses the global 'greeting'

say_greeting()     # Hello, World!
print(greeting)    # Hello — still accessible here too
````

````python
# --- Local vs Global: same name, different variables ---
# A local variable can have the same name as a global one.
# Inside the function, the LOCAL version takes priority ("shadows" the global).

color = "blue"                  # Global variable

def paint():
    color = "red"               # Local variable — different from the global 'color'
    print(f"Inside paint(): color = {color}")      # red

paint()
print(f"Outside paint(): color = {color}")         # blue — unchanged!
````

````python
# --- The 'global' keyword: modifying a global variable from inside a function ---
# Use this SPARINGLY — it can make code hard to debug.

counter = 0                     # Global variable

def increment():
    global counter              # Tell Python: I want to modify the GLOBAL 'counter'
    counter += 1

print(f"Before: counter = {counter}")    # 0
increment()
increment()
increment()
print(f"After:  counter = {counter}")    # 3
````

````python
# --- ✅ Better approach: avoid 'global' — use parameters and return values ---
# This is cleaner, easier to test, and less error-prone.

def increment_value(current_count):
    """Return the incremented count without touching any global."""
    return current_count + 1

counter = 0
counter = increment_value(counter)    # 1
counter = increment_value(counter)    # 2
counter = increment_value(counter)    # 3
print(f"Counter: {counter}")          # 3 — same result, but no global keyword needed
````

````python
# --- Scope in nested functions (bonus) ---
# Inner functions can read variables from the enclosing function.

def outer():
    message = "Hello from outer"     # Enclosing scope (not global, not local to inner)

    def inner():
        print(message)               # Can read from enclosing scope

    inner()

outer()     # Hello from outer
````

### Scope Lookup Order: LEGB Rule

When Python encounters a variable name, it searches in this order:

1. **L**ocal — inside the current function
2. **E**nclosing — in any enclosing (outer) functions
3. **G**lobal — at the module/file level
4. **B**uilt-in — Python's built-in names (`print`, `len`, `range`, etc.)

The first match wins.

````python
# --- LEGB rule in action ---

x = "global x"                   # G: Global

def outer():
    x = "enclosing x"             # E: Enclosing

    def inner():
        x = "local x"             # L: Local
        print(f"inner sees:    {x}")   # local x (L wins)

    inner()
    print(f"outer sees:    {x}")       # enclosing x

outer()
print(f"module sees:   {x}")           # global x
````

## 4. Putting It All Together — A Complete Example

````python
# --- Mini Grade Calculator ---
# This example uses everything we've learned:
# defining functions, different argument types, return values, and scope.

def calculate_average(*scores):
    """Calculate the average of any number of scores."""
    if not scores:
        return 0
    return sum(scores) / len(scores)


def get_letter_grade(average, scale="standard"):
    """
    Convert a numeric average to a letter grade.
    Uses default 'standard' scale, but can be changed to 'lenient'.
    """
    if scale == "lenient":
        thresholds = {"A": 85, "B": 75, "C": 65, "D": 55}
    else:
        thresholds = {"A": 90, "B": 80, "C": 70, "D": 60}

    for grade, cutoff in thresholds.items():
        if average >= cutoff:
            return grade
    return "F"


def generate_report(student_name, *scores, **options):
    """
    Generate a full student report.

    Args:
        student_name : str       — the student's name
        *scores      : float     — any number of exam scores
        **options    : keyword    — optional settings (e.g., scale='lenient')
    """
    scale = options.get("scale", "standard")     # Default to 'standard'

    avg = calculate_average(*scores)              # Unpack and pass to average function
    grade = get_letter_grade(avg, scale=scale)

    # Build the report
    report = {
        "student": student_name,
        "scores": scores,
        "average": round(avg, 1),
        "grade": grade,
        "scale": scale
    }
    return report


# --- Generate some reports ---
report1 = generate_report("Alice", 92, 88, 95, 91)
report2 = generate_report("Bob", 72, 68, 75, 70)
report3 = generate_report("Charlie", 72, 68, 75, 70, scale="lenient")

for report in [report1, report2, report3]:
    print(f"Student: {report['student']}")
    print(f"  Scores:  {report['scores']}")
    print(f"  Average: {report['average']}")
    print(f"  Grade:   {report['grade']} ({report['scale']} scale)")
    print()
````
