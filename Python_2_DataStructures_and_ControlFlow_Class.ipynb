# Part 1 — Data Structures in Python

A **data structure** is just a container that holds data and gives you ways to
organize and work with it. Python ships with four core built-in containers, and
almost everything you'll ever build uses some combination of them:

| Structure | Looks like | Ordered? | Changeable (mutable)? | Allows duplicates? | Typical use |
|-----------|-----------|----------|------------------------|--------------------|-------------|
| **list** | `[1, 2, 3]` | ✅ Yes | ✅ Yes | ✅ Yes | A sequence of items you'll modify |
| **tuple** | `(1, 2, 3)` | ✅ Yes | ❌ No | ✅ Yes | A fixed group of values |
| **set** | `{1, 2, 3}` | ❌ No | ✅ Yes | ❌ No | Unique items, fast membership tests |

## 1.1 — Lists

A **list** is an ordered, changeable collection.
You create one with square brackets `[ ]`.


````python
# A list can hold anything — and even mix types
fruits = ["apple", "banana", "cherry"] # list of fruits
numbers = [10, 20, 30, 40, 50] # list of numbers
mixed = ["text", 42, 3.14, True] # list can contain multiple data types

s1 = "banana"
print(len(s1))
s2 = ["banana"]
print(len(s2))

print("fruits:", fruits)
print("numbers:", numbers)
print("mixed:", mixed)

# len() tells you how many items are in a list
print("There are", len(fruits), "fruits")
````

### Indexing — grabbing one item

Each item has a position called an **index**. 🧠 **Python counts from 0**, not 1.

```
   fruits = [ "apple" , "banana" , "cherry" ]
   index:       0          1          2
   negative:   -3         -2         -1
```

Negative indexes count from the **end**, which is handy for getting the last item.


````python
fruits = ["apple", "banana", "cherry", "date"]

print(fruits[0])    # first item  -> apple
print(fruits[2])    # third item  -> cherry
print(fruits[-1])   # last item   -> date
print(fruits[-2])   # second-last -> cherry
````

### Slicing — grabbing a *range* of items

Slicing uses `list[start : stop : step]`. The **start is included** but the
**stop is excluded** (this trips up everyone at first 🧠).


````python
nums = [0, 10, 20, 30, 40, 50, 60, 70]

print(nums[2:5])    # index 2,3,4      -> [20, 30, 40]
print(nums[:3])     # from start to 3  -> [0, 10, 20]
print(nums[4:])     # from 4 to end    -> [40, 50, 60, 70]
print(nums[::2])    # every 2nd item   -> [0, 20, 40, 60]
print(nums[::-1])   # reversed!        -> [70, 60, ... 0]
print(nums[::1])
````

### Mutability — lists can change

🧠 Lists are **mutable**: you can change, add, or remove items after creating the
list. Notice we change an item *in place* by assigning to an index.


````python
colors = ["red", "green", "blue"]
print("before:", colors)

colors[1] = "GREEN"      # replace the item at index 1
print("after :", colors)

colors[1:2] = "orange", "yellow"
print("after 2:", colors)
````

### Common list operations

These are the methods you'll reach for constantly:

- `append(x)` — add **one** item to the end
- `extend([...])` — add **several** items (merge another list in)
- `insert(i, x)` — add an item at a specific position
- `pop(i)` — remove & **return** the item at index `i` (default: last)
- `remove(x)` — remove the first item equal to `x`
- `sort()` / `reverse()` — reorder the list in place


````python
basket = ["apple", "banana"]

basket.append("cherry")            # add one item
print("after append:", basket)

basket.extend(["date", "elderberry"])  # add multiple items
print("after extend:", basket)

basket.insert(0, "apricot")        # add at the front
print("after insert:", basket)
````

````python
basket = ["apricot", "apple", "banana", "cherry", "date"]

removed = basket.pop()             # removes & returns the LAST item
print("popped:", removed, "-> list:", basket)

removed = basket.pop(0)            # removes & returns item at index 0
print("popped:", removed, "-> list:", basket)

basket.remove("cherry")            # remove a specific value
print("after remove:", basket)
````

#### ⚠️ append vs extend — a classic gotcha

`append` adds its argument as a **single element**. If you `append` a list, you get
a *list inside a list*. `extend` unpacks the items instead. Run this to see it:


````python
a = [1, 2, 3]
a.append([4, 5])     # the whole list becomes ONE element
print("append:", a)  # -> [1, 2, 3, [4, 5]]

b = [1, 2, 3]
b.extend([4, 5])     # each item added separately
print("extend:", b)  # -> [1, 2, 3, 4, 5]
````

````python
# Sorting and reversing
scores = [42, 17, 99, 3, 58]
scores.sort()
print("sorted:", scores)

scores.sort(reverse=True)
print("sorted desc:", scores)

names = ["Charlie", "alice", "Bob"]
print("sorted names:", sorted(names))   # sorted() returns a NEW list
````

> 🧠 **`sort()` vs `sorted()`** — `mylist.sort()` changes the list in place and
> returns `None`. `sorted(mylist)` leaves the original alone and returns a brand-new
> sorted list. A very common bug is writing `x = mylist.sort()` and finding `x` is
> `None`!

## 1.2 — Tuples

A **tuple** is like a list, but **immutable** — once created, it cannot be changed.
You create one with parentheses `( )`. Use a tuple when a group of values belongs
together and should *not* be modified (coordinates, RGB colors, database records).


````python
point = (3, 7)
rgb = (255, 128, 0)
person = ("Ada", "Lovelace", 1815)

print("point:", point)
print("rgb:", rgb)
print("first name:", person[0])   # indexing works just like lists
print("slice:", rgb[0:2])         # slicing works too
````

### Immutability in action

Trying to change a tuple raises an error. Run this and read the error message —
learning to read errors is a real skill 🧠.


````python
point = (3, 7)

try:
    point[0] = 99          # not allowed!
except TypeError as e:
    print("Error:", e)
````

### Packing & unpacking

Tuples shine at **unpacking** — assigning several variables at once. This is used
constantly in real Python code (and is how you swap variables without a temp var).


````python
# Packing: group values into a tuple
record = ("Ada", 36, "Engineer")

# Unpacking: split them back out into variables
name, age, role = record
print(name, "is a", age, "year-old", role)
````

> 🧠 **Single-item tuples need a comma:** `(5)` is just the number 5, but `(5,)`
> is a tuple. The comma — not the parentheses — is what makes a tuple.


````python
not_a_tuple = (5)
yes_a_tuple = (5,)
lst = []
print(type(not_a_tuple))   # <class 'int'>
print(type(yes_a_tuple))   # <class 'tuple'>
print(type(lst))
````

## 1.3 — Sets

A **set** is an **unordered** collection of **unique** items. You create one with
curly braces `{ }`. Sets automatically throw away duplicates and are extremely fast
at answering "is this item in here?".


````python
# Duplicates vanish automatically
numbers = {1, 2, 2, 3, 3, 3, 4}
print("set:", numbers)          # -> {1, 2, 3, 4}

# A classic trick: remove duplicates from a list
names = ["Ana", "Bob", "Ana", "Cara", "Bob"]
unique_names = set(names)
print("unique:", unique_names)
print("count of unique:", len(unique_names))
````

> ⚠️ To make an **empty set** you must use `set()`, not `{}` — because `{}`
> creates an empty **dictionary**!


### Changing a set: add, update, remove, discard

- `add(x)` — add one item
- `update([...])` — add many items
- `remove(x)` — remove `x`, **error** if missing
- `discard(x)` — remove `x`, **no error** if missing


````python
colors = {"red", "green"}

colors.add("blue")                 # add one
colors.update(["yellow", "cyan"])  # add several
print("after adds:", colors)

colors.discard("pink")  # not there — no error
colors.remove("red")    # remove an existing item
print("after removes:", colors)
````

### Set math — the real superpower

Sets let you compute **unions, intersections, and differences** — perfect for
comparing two groups of things.


````python
math_students = {"Ana", "Bob", "Cara", "Dan"}
art_students  = {"Cara", "Dan", "Eve", "Finn"}

print("In either class (union)       :", math_students | art_students)
print("In BOTH classes (intersection):", math_students & art_students)
print("Only in math (difference)     :", math_students - art_students)
print("In exactly one (symmetric)    :", math_students ^ art_students)

# Membership testing is very fast
print("Is Bob in math?", "Bob" in math_students)
````

## 1.4 — Dictionaries

A **dictionary** stores data as **key → value** pairs. Instead of looking items up
by position (like a list), you look them up by a meaningful **key**. This is the
structure you'll use to represent "things with properties" — a user, a product, a
config.


````python
student = {
    "name": "Ada",
    "age": 19,
    "grades": [88, 92, 79],
    "active": True,
}

print(student)
print("Name:", student["name"])      # look up by key
print("First grade:", student["grades"][0])
````

### Adding, updating, and removing

- Assign to a new key to **add** it: `d["key"] = value`
- Assign to an existing key to **overwrite** it
- `update({...})` merges another dict in (add/overwrite several at once)
- `pop("key")` removes a key and returns its value


````python
student = {"name": "Ada", "age": 19}

student["email"] = "ada@example.com"   # add a new key
student["age"] = 20                    # update existing
student.update({"age": 21, "city": "London"})  # change + add together
print("updated:", student)

removed = student.pop("email")         # remove and return
print("removed email:", removed)
print("final:", student)
````

### Looping over a dictionary

`.keys()`, `.values()`, and `.items()` let you iterate over the parts of a dict.
`.items()` gives you both the key and value together — you'll use it constantly.


````python
prices = {"coffee": 3.5, "tea": 2.0, "juice": 4.25}

print("Keys:")
for item in prices.keys():
    print("  -", item)

print("Values:")
for cost in prices.values():
    print("  -", cost)

print("Both together:")
for item, cost in prices.items():
    print(f"  {item} costs ${cost}")
````

> 🧠 Keys must be **unique** and **immutable** (strings, numbers, tuples work as
> keys; lists do not). Values can be anything at all — including other dicts and
> lists, which is how you build nested data.

# Part 2 — Control Flow

So far our code has run top-to-bottom, one line after another. **Control flow** lets

where code starts to feel alive.


## 2.1 — Conditional statements: `if`, `elif`, `else`

An `if` statement runs a block of code **only when a condition is true**.

🧠 Two things every beginner must internalise:
- The condition line ends with a **colon** `:`
- The code that belongs to the `if` is **indented** (4 spaces). Indentation is not
  decoration in Python — it *is* the syntax that groups statements.


````python
temperature = 24

if temperature > 23:
    print("It's hot!")
    print("Stay hydrated. 💧")

print("This line always runs (it's not indented).")
````

### Comparison & logical operators

Conditions are built from these:

| Operator | Meaning | Example |
|----------|---------|---------|
| `==` | equal to | `x == 5` |
| `!=` | not equal | `x != 5` |
| `>` `<` | greater / less than | `x > 5` |
| `>=` `<=` | greater/less or equal | `x >= 5` |
| `and` | both must be true | `x > 0 and x < 10` |
| `or` | at least one true | `x < 0 or x > 100` |
| `not` | flips true/false | `not done` |

> ⚠️ `=` assigns a value; `==` compares two values. Mixing these up is the #1
> beginner bug.


````python
age = 16
has_ticket = True

if age >= 18 and has_ticket:
    print("Welcome to the show!")
else:
    print("Sorry, you can't enter.")
````

### `elif` — choosing among many options

When you have more than two outcomes, chain conditions with `elif` ("else if").
Python checks them **top to bottom** and runs the **first** one that's true, then
skips the rest.


````python
score = 83

if score >= 90:
    grade = "A"
elif score >= 80:
    grade = "B"
elif score >= 70:
    grade = "C"
elif score >= 60:
    grade = "D"
else:
    grade = "F"

print("Score", score, "earns grade", grade)
````

### Nesting conditionals

You can put an `if` inside another `if` when a decision depends on a previous one.
Keep an eye on the indentation levels.


````python
logged_in = True
is_admin = False

if logged_in:
    print("Welcome back!")
    if is_admin:
        print("You have admin access.")
    else:
        print("You have standard access.")
else:
    print("Please log in first.")
````

## 2.2 — Loops

A **loop** repeats a block of code. Python has two kinds:
- **`for`** — repeat once for each item in a collection (most common)
- **`while`** — repeat *as long as* a condition stays true


### `for` loops

A `for` loop walks through any collection — a list, a string, a set, a dict — one
item at a time.


````python
fruits = ["apple", "banana", "cherry"]

for n in fruits:
    print("I like", n)

print("---")

# Looping over a string gives you each character
for letter in "Hi!":
    print(letter)
````

### `range()` — looping a set number of times

`range(stop)` produces numbers `0, 1, ... stop-1`. You can also give it a start and
a step: `range(start, stop, step)`.


````python
numbers = range(10)
numbers
````

````python
# Print 0 through 4
for i in range(5):
    print("count:", i)
````

````python
# Print 0 through 4
for i in range(5):
    print("count:", i)

print("---")

# Start at 2, stop before 10, step by 2
for n in range(2, 10, 2):
    print("even:", n)

print("---")

# Count down
for n in range(5, 0, -1):
    print("T-minus", n)
````

> 🧠 **`enumerate()`** gives you the index *and* the item at the same time —
> cleaner than manually tracking a counter.


````python
teams = ["Red", "Green", "Blue"]
enumerate(teams)
````

````python
teams = ["Red", "Green", "Blue"]

for index, team in enumerate(teams):
    print(index, "->", team)
````

````python
teams = ["Red", "Green", "Blue"]

for index in range(len(teams)):
    team = teams[index]
    print(index, "->", team)
````

### `while` loops

A `while` loop keeps going until its condition becomes false. ⚠️ You **must** make
sure the condition eventually becomes false, or you'll create an *infinite loop*
(in Colab, press the ⏹ stop button to interrupt one).


````python
countdown = 3
while countdown > 0:
    print("Counting:", countdown)
    countdown = countdown - 1   # this line is what eventually stops the loop
print("Lift off! 🚀")
````

### `break`, `continue`, and `pass`

These three keywords give you fine control inside loops:

- **`break`** — exit the loop **immediately**
- **`continue`** — skip to the **next iteration**, ignoring the rest of this one
- **`pass`** — do **nothing** (a placeholder so the block isn't empty)


````python
# break: stop as soon as we find the target
for n in [1, 3, 5, 8, 9, 10]:
    if n % 2 == 0:
        print("Found the first even number:", n)
        break
    print("checking", n, "...")
````

````python
# continue: skip the even numbers, print only the odds
for n in range(1, 11):
    if n % 2 == 0:
        continue          # skip the rest of the loop body for even n
    print("odd:", n)
````

````python
# pass: a do-nothing placeholder (useful while sketching code)
for n in range(3):
    if n == 1:
        pass              # "I'll fill this in later"
    else:
        print("handling", n)
````

### Combining loops with data structures

Loops and dictionaries are a perfect pair. Here we tally up word counts — a real
task you'll do often.


````python
sentence = "the cat sat on the mat the cat ran"
words = sentence.split()
words
````

````python
sentence = "the cat sat on the mat the cat ran"
words = sentence.split()        # split into a list of words

counts = {}                     # empty dict to tally results
for word in words:
    if word in counts:
        counts[word] = counts[word] + 1
    else:
        counts[word] = 1

print(counts)
````

## 2.3 — List comprehensions

A **list comprehension** is a compact, readable way to build a list from a loop in a
single line. Once it clicks, you'll use it everywhere.

**The pattern:**

```
[ expression  for item in iterable ]
   │             │
   │             └── the loop
   └── what to do with each item
```


````python
# The long way
squares = []
for n in range(1, 6):
    squares.append(n ** 2)
print("loop  :", squares)

# The comprehension way — same result, one line
squares = [n ** 2 for n in range(1, 6)]
print("comp  :", squares)
````

### Adding a condition (filtering)

Tack an `if` onto the end to keep only items that match.

```
[ expression  for item in iterable  if condition ]
```


````python
numbers = range(1, 11)

evens = [n for n in numbers if n % 2 == 0]
print("evens:", evens)

# Transform AND filter at once: squares of the odd numbers
odd_squares = [n ** 2 for n in numbers if n % 2 == 1]
print("odd squares:", odd_squares)
````

````python
# Works on any iterable — here we clean up a list of words
words = ["  Apple ", "BANANA", "cherry  "]
clean = [w.strip().lower() for w in words]
print(clean)   # -> ['apple', 'banana', 'cherry']
````

### Conditional *expression* inside a comprehension

You can also choose between two values per item using `A if cond else B` at the
**front** of the comprehension. (Note this is different from the filtering `if` at
the end.)


````python
numbers = [1, 2, 3, 4, 5, 6]
labels = ["even" if n % 2 == 0 else "odd" for n in numbers]
print(labels)   # -> ['odd', 'even', 'odd', 'even', 'odd', 'even']
````

### Set and dictionary comprehensions

The same idea works with `{ }` to build sets and dicts.


````python
# Set comprehension — unique word lengths
words = ["hi", "bye", "yes", "no", "okay"]
lengths = {len(w) for w in words}
print("unique lengths:", lengths)

# Dict comprehension — map each word to its length
length_map = {w: len(w) for w in words}
print("length map:", length_map)
````

> 🧠 **Readability matters.** Comprehensions are great for simple transforms and
> filters. If the logic gets complicated (multiple nested conditions, side effects),
> a regular `for` loop is clearer — and clearer is always better.

## ✅ Recap — what you learned

**Data structures**
- **Lists** `[ ]` — ordered, mutable; `append`, `extend`, `insert`, `pop`, `remove`, `sort`
- **Tuples** `( )` — ordered, immutable; great for fixed groups and unpacking
- **Sets** `{ }` — unordered, unique items; `add`, `update`, `remove`, and set math (`|`, `&`, `-`)
- **Dictionaries** `{key: value}` — labeled data; `get`, `update`, `pop`, `.items()`
- **Indexing & slicing** — `x[0]`, `x[-1]`, `x[start:stop:step]`
- **Mutability** — lists/sets/dicts can change; tuples cannot

**Control flow**
- **Conditionals** — `if` / `elif` / `else` with comparison & logical operators
- **Loops** — `for` (with `range`, `enumerate`) and `while`
- **`break` / `continue` / `pass`** — fine control inside loops
- **List comprehensions** — `[expr for item in iterable if cond]`, plus set & dict versions
